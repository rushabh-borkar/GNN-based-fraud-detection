##
1. Classes
The classes file contains the transaction IDs and their corresponding labels. The dataset is highly imbalanced, with illicit transactions representing approximately 9.76% of the labeled transactions. Unknown transactions have no known class label and should not be treated as either licit or illicit.
2. Features
The features file contains 165 numerical features for each transaction, along with its transaction ID and time step. The time step ranges from 1 to 49 and represents the transaction's temporal interval in the dataset. The features are already largely standardized, and no missing values were found.
3. Edgelist
The edgelist represents relationships between transactions. These connections are essential for constructing a graph, allowing a Graph Neural Network (GNN) to learn from both transaction attributes and neighboring transactions. Its structure and consistency still need to be validated.

##TRYING TO BALANCE THE TRAINING DATA

```python
labeled_df = df_classes[df_classes['class'] != 'unknown']
X = labeled_df.iloc[: , 2:]
y = labeled_df['class']
```

```python
labeled_df = df_classes[df_classes['class'] != 'unknown']

# Merge labels with features using transaction ID
df_features = df_features.rename(columns={0: 'txId'})

merged_df = labeled_df.merge(df_features, on='txId')

# Define features and target
X = merged_df.iloc[:, 3:]  # 165 features, excluding txId and time step
y = merged_df['class']

# Split the data
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.2,
    stratify=y,
    random_state=42
)

print(X_train.shape)
print(X_test.shape)
print(y_train.shape)
print(y_test.shape)
```

```python
print("Training set:")
print(y_train.value_counts())
print((y_train.value_counts(normalize=True) * 100).round(2))

print("\nTesting set:")
print(y_test.value_counts())
print((y_test.value_counts(normalize=True) * 100).round(2))
```

##Transaction ID to Node Index Mapping

```python
txt_id = df_features['txId']

print("Total transaction: ", len(txt_id))
print("Unique transactions: ", txt_id.nunique())
print("The Duplicates transaction: ",txt_id.duplicated().sum())
```

```python
# Map transaction IDs to node indices
node_mapping = {
    tx_id: idx
    for idx, tx_id in enumerate(txt_id)
}
```

```python
print("The length: ",len(node_mapping))

print(list(node_mapping.items())[:5])
```

```python
##Reverse mapping for retriving the data
reverse_mapping = {
    idx: tx_id
    for tx_id, idx in node_mapping.items()
}
```

```python
num_nodes = len(node_mapping)

assert num_nodes == len(df_features), "Node count mismatch!"
assert len(set(node_mapping.values())) == num_nodes, "Duplicate node indices!"

print("Node mapping successful!")
```

##Check transaction IDs against mapping

```python
missing_source = ~df_edgelist['txId1'].isin(node_mapping)

missing_destination = ~df_edgelist['txId2'].isin(node_mapping)

print("Missing source IDs:", missing_source.sum())
print("Missing destination IDs:", missing_destination.sum())
```

Converting source transaction IDs and  Converting destination transaction IDs

```python
df_edgelist['source_idx'] = df_edgelist['txId1'].map(node_mapping)
df_edgelist['destination_idx'] = df_edgelist['txId2'].map(node_mapping)
```
